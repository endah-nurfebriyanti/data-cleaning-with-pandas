```python
import pandas as pd
```

```python
df=pd.read_csv("Harvard_English_Centre_Payments_2025_Dirty.csv")
```

```python
df=df.drop_duplicates()
```

```python
df.isnull().sum()
```

```python
df["Tutor"]=df["Tutor"].fillna("Unknown")
```

```python
df["Discount"]=df["Discount"].fillna(0)
```

```python
df["Amount_Paid"]=df["Amount_Paid"].fillna(df["Course_Fee"]-df["Discount"])
```

```python
df.isnull().sum()
```

```python
df["Payment_Method"]=df["Payment_Method"].fillna("Unknown")
```

```python
df.isnull().sum()
```

```python
df["Payment_Date"]=pd.to_datetime(df["Payment_Date"],format="mixed",dayfirst=True)
```

```python
df["Year"]=df["Payment_Date"].dt.year
```

```python
df["Quarter"]=df["Payment_Date"].dt.quarter
```

```python
df["Month_Name"]=df["Payment_Date"].dt.month_name()
```

```python
df["Week"]=df["Payment_Date"].dt.isocalendar().week
```

```python
df["Day"]=df["Payment_Date"].dt.day_name()
```

```python
df.info()
```

```python
df.isnull().sum()
```

```python
df.duplicated().sum()
```

```python
df.to_csv("Harvard_English_Data_Clean.csv",index=False)
```
